```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/financial_loan.csv", encoding="cp1252")

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
date_columns = [
    "issue_date",
    "last_credit_pull_date",
    "last_payment_date",
    "next_payment_date"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")
```

```python
df.describe()
```

```python
loan_status_counts = df["loan_status"].value_counts()

loan_status_counts
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="loan_status",
    color="#A8C3B0"
)

plt.title("Loan Status Distribution", fontsize=14)
plt.xlabel("Loan Status")
plt.ylabel("Number of Loans")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
loan_status_counts = df["loan_status"].value_counts()

loan_status_counts
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="loan_status",
    color="#A8C3B0"
)

plt.title("Loan Status Distribution", fontsize=14)
plt.xlabel("Loan Status")
plt.ylabel("Number of Loans")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
purpose_counts = df["purpose"].value_counts().head(10)

purpose_counts
```

```python
plt.figure(figsize=(9, 5))

purpose_counts.sort_values().plot(
    kind="barh",
    color="#A9C4D4"
)

plt.title("Top Loan Purposes", fontsize=14)
plt.xlabel("Number of Applications")
plt.ylabel("Loan Purpose")
plt.tight_layout()
plt.show()
```

```python
grade_interest = (
    df.groupby("grade")["int_rate"]
      .mean()
      .sort_index()
)

grade_interest
```

```python
plt.figure(figsize=(8, 5))

grade_interest.plot(
    kind="bar",
    color="#C8B6D8"
)

plt.title("Average Interest Rate by Loan Grade", fontsize=14)
plt.xlabel("Loan Grade")
plt.ylabel("Average Interest Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
dti_by_status = (
    df.groupby("loan_status")["dti"]
      .mean()
      .sort_values(ascending=False)
)

dti_by_status
```

```python
plt.figure(figsize=(8, 5))

dti_by_status.plot(
    kind="bar",
    color="#E8B4A0"
)

plt.title("Average Debt-to-Income Ratio by Loan Status", fontsize=14)
plt.xlabel("Loan Status")
plt.ylabel("Average DTI")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```
